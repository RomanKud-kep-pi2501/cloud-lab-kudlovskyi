### Завдання 4. Набір даних (CI/CD-збірки)
Дані навчальні, синтетичні.

```python
import pandas as pd
import numpy as np

# Фіксуємо seed для відтворюваності результатів
np.random.seed(14)

n_rows = 35

pipelines = np.random.choice(
    ['build-backend', 'build-frontend', 'deploy-staging', 'run-tests', 'lint-check'],
    size=n_rows
)
branches = np.random.choice(
    ['main', 'develop', 'feature/auth', 'feature/api', 'hotfix/login'],
    size=n_rows,
    p=[0.35, 0.30, 0.15, 0.10, 0.10]
)
# Тривалість збірки в секундах (правдоподібний діапазон від 25 до 480 с)
duration_s = np.random.randint(25, 481, size=n_rows)
# Результат збірки (success / failed)
result = np.random.choice(['success', 'failed'], size=n_rows, p=[0.78, 0.22])
# Дати збірок за листопад 2026
dates = pd.date_range(start='2026-11-01', periods=n_rows, freq='6h').strftime('%Y-%m-%d %H:%M')

df = pd.DataFrame({
    'pipeline': pipelines,
    'branch': branches,
    'duration_s': duration_s,
    'result': result,
    'date': dates
})

# Перевірка даних на дублікати та коректність діапазонів
print("Кількість повних дублікатів:", df.duplicated().sum())
print("Мінімальна та максимальна тривалість (с):", df['duration_s'].min(), "-", df['duration_s'].max())

# Збереження у CSV-файл
df.to_csv('cicd_builds_v14.csv', index=False)

# Виведення основної інформації про DataFrame
print("\nРозмір таблиці (df.shape):", df.shape)
print("\nПерші 5 рядків (df.head()):")
display(df.head())

print("\nІнформація про стовпці та типи даних (df.info()):")
df.info()
```
